## Reminder: Special Methods

| Method | Called by |
|:--|:--|
| `__init__()` | instantiation, e.g. `Person('Greta', 'Thunberg')` |
| `__str__()` | `print()` and `str()` |
| `__lt__()` | the operator `<` |
| `__le__()` | the operator `<=` |
| `__eq__()` | the operator `==` |
| `__ne__()` | the operator `!=` (defaults to the opposite of `__eq__()`) |
| `__gt__()` | the operator `>` |
| `__ge__()` | the operator `>=` |

## Reminder: The Class `Person`

```python
from datetime import date

class Person(object):
    """A class to represent a person."""
        
    def __init__(self, f_name, l_name):
        """Creates a person using first and last names."""
        self.first_name = f_name
        self.last_name = l_name
        self.birthdate = None
    
    def get_name(self):
        """Gets self's full name."""
        return self.first_name + ' ' + self.last_name
    
    def get_age(self):
        """Gets self's age in years."""
        return date.today().year - self.birthdate.year
    
    def set_birthdate(self, dob):
        """Assumes dob is of type date.
        Sets self's birthdate to dob.
        """
        self.birthdate = dob
    
    def __str__(self):
        """Returns self's full name."""
        return self.get_name()
    
    def __lt__(self, other):
        """Returns True if self's last name precedes other's last name
        in alphabetical order. If they are equal, compares first names.
        """
        if self.last_name == other.last_name:
            return self.first_name < other.first_name
        return self.last_name < other.last_name

p1 = Person('Greta', 'Thunberg')
p2 = Person('Tom', 'Holland')
p1.set_birthdate(date(2003, 1, 3))
print(p1, p1.get_age(), p1 < p2)
```

## Overriding a Method

* Here, `LSEPerson` inherits `__lt__()` from `Person`, so LSE people are compared by name
* We want to compare them by ID number instead
* → We can **override** an inherited method by defining a method with the same name

## Adding the Comparison Method

```python
class LSEPerson(Person):
    """A class to represent a student or staff member at LSE."""
    next_id_num = 1 # class variable: the next unique ID number

    def __init__(self, f_name, l_name, department=None):
        """Creates an LSE person using names and a department."""
        super().__init__(f_name, l_name)  # calls Person.__init__
        self.department = department
        self.id_num = LSEPerson.next_id_num
        LSEPerson.next_id_num += 1

    def get_department(self):
        """Gets self's department."""
        return self.department

    def get_id_num(self):
        """Gets self's unique LSE number."""
        return self.id_num

    def __lt__(self, other):
        """Returns True if self's id number is smaller than other's."""
        return self.id_num < other.id_num
```

## Comparing LSE People

`a < b` is shorthand for `a.__lt__(b)`, so the object on the **left** decides which version runs.

```python
# new objects: the old ones still belong to the old class
staff1 = LSEPerson('Oliver', 'Rittmann', 'Methodology')
staff2 = LSEPerson('Ed', 'Kelly', 'Methodology')

print(staff1 < staff2)  # LSEPerson.__lt__: compares id numbers
print(p1 < staff1)      # Person.__lt__: compares names
```

```python
print(staff1 < p1)      # LSEPerson.__lt__: but p1 has no id number!
```

## Extending a Method

A subclass can also **build on** the method it overrides: `super()` gives access to the version of the superclass.

```python
class Visitor(LSEPerson):
    """A class to represent a visitor at LSE."""

    def __str__(self):
        """Returns self's full name, marked as a visitor."""
        return super().__str__() + ' (visitor)'

visitor1 = Visitor('Tom', 'Holland')
print(visitor1)
```

## Why an Intermediate Class?

Two ways to check whether someone is a student. We add a new kind of student later: which one still works?

<div class="code-columns"></div>

```python
def is_student_v1(person):
    return (type(person) == Undergrad
            or type(person) == Grad)
```

```python
def is_student_v2(person):
    return isinstance(person, Student)
```

```python
class TransferStudent(Student):
    def __init__(self, f_name, l_name, from_school):
        super().__init__(f_name, l_name)
        self.from_school = from_school

student1 = TransferStudent('Jane', 'Doe', 'MIT')
print(is_student_v1(student1), is_student_v2(student1))
```

Good programmers design their programs so that as little code as possible has to change when new classes are added.

## The Substitution Principle

# Encapsulation and Information Hiding

## A Class That Uses Our Hierarchy

<div style="box-sizing: border-box; margin: 0px 0 4px 0; padding: 12px 18px; border-radius: 8px; background-color: #f7efe6; font-family: system-ui; line-height: 1.45; color: #212121;">Now other code uses our classes. What can go wrong?</div>

```python
class Grades(object):
    """A grade book for one course."""

    def __init__(self):
        self.students = []
        self.grades = {}  # maps id numbers to lists of grades

    def add_student(self, student):
        self.students.append(student)
        self.grades[student.get_id_num()] = []

    def add_grade(self, student, grade):
        self.grades[student.get_id_num()].append(grade)

    def get_grades(self, student):
        return self.grades[student.get_id_num()]

    def get_students(self):
        return self.students
```

## The Problem: Getting the List Itself

```python
course = Grades()
course.add_student(Grad('Jane', 'Doe'))
course.add_student(Undergrad('John', 'Doe'))

all_students = course.get_students()
all_students.append(Grad('Liz', 'Beaver'))

for student in course.get_students():
    print(student)
```

```python
# predict, then check: what does course.get_grades(all_students[-1]) do?
```

## The Fix: Return a Copy

```python
class SafeGrades(Grades):
    """A grade book that protects its list of students."""
    def get_students(self):
        return self.students[:]  # a copy of the list

course = SafeGrades()
course.add_student(Grad('Jane', 'Doe'))

all_students = course.get_students()
all_students.append(Grad('Liz', 'Beaver'))
print(len(course.get_students()))
```

A copy protects the **list**, not the objects in it:

```python
all_students[0].first_name = 'Changed'  # for demonstration only
print(course.get_students()[0])
```

## Information Hiding in Python

* Use naming conventions to make data attributes and methods invisible outside the class
* Convention: begin the name with `__`, but do not end it with `__`

```python
class InfoHiding(object):
    def __init__(self):
        self.visible = 'Look at me'
        self.__visible__ = 'Look at me too'
        self.__invisible = 'Do not look at me directly'
        
    def print_invisible(self):
        print(self.__invisible)
        
    def __invisible_print_invisible(self):
        print(self.__invisible)
        
    def __visible_print_invisible__(self):
        print(self.__invisible)
```

## Information Hiding in Python: Accessing Attributes

```python
test = InfoHiding()

# These are for demonstration purposes only!
# You should not access object attributes directly like this, 
# use get and set methods instead
print(test.visible)
print(test.__visible__)
print(test.__invisible)
```

The name is not really hidden: Python renames the attribute, so it can still be reached.

```python
print(test._InfoHiding__invisible)  # for demonstration only
```

## Information Hiding in Python: Calling Methods

```python
test.print_invisible()
test.__visible_print_invisible__()
test.__invisible_print_invisible()
```

## Information Hiding and Subclasses

The subclass cannot see the attribute either: Python looks for the renamed `_SubClass__invisible`. This makes `__` cumbersome.

```python
class SubClass(InfoHiding):
    def __init__(self):
        InfoHiding.__init__(self)
        print(self.__invisible)

sub_test = SubClass()
```

## ⚡️ Generators with `yield`

A function that contains `yield` is a **generator**: it hands out values one at a time, and no new list is created. The `for` loop stays the same.

```python
class FastGrades(Grades):
    """A grade book that hands out its students one at a time."""

    def get_students(self):
        for student in self.students:
            yield student

course = FastGrades()
course.add_student(Grad('Julie', 'Doe'))
course.add_student(Grad('Lisa', 'Doe'))

print(course.get_students())
print(sys.getsizeof(course.get_students()), 'bytes')
```

## How a Generator Works

* `for` starts the generator, which runs until the first `yield` and hands out that value
* In the next iteration, it **resumes right after** the `yield`, with its local variables intact
* The loop ends when the generator runs out of code

<div class="code-right"></div>

```python
def count_to_two():
    print('starting')
    yield 1
    print('resuming')
    yield 2
    print('finished')

for number in count_to_two():
    print(number)
```
