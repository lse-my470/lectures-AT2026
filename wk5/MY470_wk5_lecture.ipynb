# Defining New Classes

## Creating a New Class

The following code defines a new class `Person`. It only has the special method `__init__()`, the **constructor**, which runs when we create a new object (an **instance**) and stores the object's **data attributes**:

````python
class Person(object):
    """A class to represent a person."""

    def __init__(self, f_name, l_name):
        """Creates a person using first and last names."""
        self.first_name = f_name
        self.last_name = l_name
        self.birthdate = None
````

````python
p1 = Person('Greta', 'Thunberg')   # creates an instance: calls __init__()
print(vars(p1))                    # the data attributes stored in p1
print(p1)
````

`Person('Greta', 'Thunberg')` creates a new object, and `__init__()` stores its three data attributes (`self` is the new object). `print(p1)` only tells us where `p1` is stored in memory; we fix this later with `__str__()`.

## Classes vs. Objects

* `Person` is a class
* `p1` is an instance of the class `Person`; it is an object of type `Person`
* Similarly, `str` is a class and `'Greta Thunberg'` is an object of type `str`

## Printing a Person

`print(p1)` only showed where `p1` is stored in memory. The special method `__str__()` defines what `print()` and `str()` show:

````python
class Person(object):
    """A class to represent a person."""

    def __init__(self, f_name, l_name):
        """Creates a person using first and last names."""
        self.first_name = f_name
        self.last_name = l_name
        self.birthdate = None

    def get_name(self):
        """Gets self's full name."""
        return self.first_name + ' ' + self.last_name

    def get_age(self):
        """Gets self's age in years."""
        return date.today().year - self.birthdate.year

    def set_birthdate(self, dob):
        """Assumes dob is of type date.
        Sets self's birthdate to dob.
        """
        self.birthdate = dob

    def __str__(self):  # new
        """Returns self's full name."""
        return self.get_name()

p1 = Person('Greta', 'Thunberg')
print(p1)
````

## Special Methods: An Example

````python
class Person(object):
    """A class to represent a person."""
        
    def __init__(self, f_name, l_name):
        """Creates a person using first and last names."""
        self.first_name = f_name
        self.last_name = l_name
        self.birthdate = None
    
    def get_name(self):
        """Gets self's full name."""
        return self.first_name + ' ' + self.last_name
    
    def get_age(self):
        """Gets self's age in years."""
        return date.today().year - self.birthdate.year
    
    def set_birthdate(self, dob):
        """Assumes dob is of type date.
        Sets self's birthdate to dob.
        """
        self.birthdate = dob
    
    def __str__(self):
        """Returns self's full name."""
        return self.get_name()
    
    def __lt__(self, other):
        """Returns True if self's last name precedes other's last name
        in alphabetical order. If they are equal, compares first names.
        """
        if self.last_name == other.last_name:
            return self.first_name < other.first_name
        return self.last_name < other.last_name
    
p1 = Person('Greta', 'Thunberg')
p2 = Person('Tom', 'Holland')
print(p1 < p2)

lst = sorted([p1, p2])
print([str(i) for i in lst])
````

## Special Methods: Another Example

Are two people with the same name equal? Without `__eq__()`, `==` only checks whether both variables refer to the **same object**:

````python
p3 = Person('Greta', 'Thunberg')
print(p1 == p3)    # two different objects
print(p1 == p1)    # the same object
````

With `__eq__()`, we decide what *equal* means for our class:

````python
class Person(object):
    """A class to represent a person."""
        
    def __init__(self, f_name, l_name):
        """Creates a person using first and last names."""
        self.first_name = f_name
        self.last_name = l_name
        self.birthdate = None
    
    def get_name(self):
        """Gets self's full name."""
        return self.first_name + ' ' + self.last_name
    
    def get_age(self):
        """Gets self's age in years."""
        return date.today().year - self.birthdate.year
    
    def set_birthdate(self, dob):
        """Assumes dob is of type date.
        Sets self's birthdate to dob.
        """
        self.birthdate = dob
    
    def __str__(self):
        """Returns self's full name."""
        return self.get_name()
    
    def __lt__(self, other):
        """Returns True if self's last name precedes other's last name
        in alphabetical order. If they are equal, compares first names.
        """
        if self.last_name == other.last_name:
            return self.first_name < other.first_name
        return self.last_name < other.last_name

    def __eq__(self, other):  # new
        """Returns True if self and other have the same first and last name."""
        return self.first_name == other.first_name and self.last_name == other.last_name

p1 = Person('Greta', 'Thunberg')
p2 = Person('Tom', 'Holland')
p3 = Person('Greta', 'Thunberg')
print(p1 == p3)
print(p1 == p2)
print(p1 != p2)    # __ne__() is the opposite of __eq__()
````

````python
class Person(object):
    """A class to represent a person."""
        
    def __init__(self, f_name, l_name):
        """Creates a person using first and last names."""
        self.first_name = f_name
        self.last_name = l_name
        self.birthdate = None
    
    def get_name(self):
        """Gets self's full name."""
        return self.first_name + ' ' + self.last_name
    
    def get_age(self):
        """Gets self's age in years."""
        return date.today().year - self.birthdate.year
    
    def set_birthdate(self, dob):
        """Assumes dob is of type date.
        Sets self's birthdate to dob.
        """
        self.birthdate = dob
    
    def __str__(self):
        """Returns self's full name."""
        return self.get_name()
    
    def __lt__(self, other):
        """Returns True if self's last name precedes other's last name
        in alphabetical order. If they are equal, compares first names.
        """
        if self.last_name == other.last_name:
            return self.first_name < other.first_name
        return self.last_name < other.last_name

    def __eq__(self, other):
        """Returns True if self and other have the same first and last name."""
        return self.first_name == other.first_name and self.last_name == other.last_name

    # add __gt__() here


p1 = Person('Greta', 'Thunberg')
p1.set_birthdate(date(2003, 1, 3))
p2 = Person('Tom', 'Holland')
p2.set_birthdate(date(1996, 6, 1))

# print(p1 > p2)    # should print False: Greta is younger than Tom
````
