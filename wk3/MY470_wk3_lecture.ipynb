## Slicing

`seq[start:end:step]` gets the elements from `start` up to, **but not including**, `end`; negative indices count from the end:

````python
s = 'Python'
````

<pre style="line-height: 1.2; letter-spacing: 0;"><code style="line-height: 1.2; letter-spacing: 0;">index       0   1   2   3   4   5

````python
s = 'Python'
print(s[0], s[-1])   # first and last element
print(s[1:4])        # indices 1, 2, 3 (4 is excluded)
print(s[:2], s[2:])  # start or end can be left out
print(s[::2])        # every second element
print(s[::-1])       # step -1: reversed
````

## Ordered vs. Unordered

In **ordered** types (sequences), every element has a position, so order matters. **Unordered** types have no positions:

````python
# Ordered: the order of elements matters
print([1, 2, 3] == [3, 2, 1])  # list
print({1, 2, 3} == {3, 2, 1})  # set
````

````python
# Only ordered types can be subset by index
ls = [10, 20, 30]
st = {10, 20, 30}
print(ls[0])   # list: index 0 is the first element
print(st[0])   # set: there is no first element
````

## Mutable vs. Immutable

**Mutable** objects can be changed in place. **Immutable** objects cannot: operations on them return a new object:

````python
# Mutable: can be changed in place
ls = ['a', 'b']
ls[0] = 'z'        # list: changed
s = 'ab'
s.upper()          # str: new string
print(ls, s)
````

````python
# Danger: two names for the same mutable object (aliasing)
a = [1, 2, 3]
b = a          # b is not a copy, just a second name (b = a[:] would copy)
b.append(4)
print(a)       # a has changed too!
````

## So Far, We've Written Straight-Line Programs

<div class="line-numbers"></div>

A **straight-line program** executes one statement after another, in the order in which they appear, and stops when it runs out of statements: every statement is executed exactly once.

_Example:_

````python
s = 'All animals are equal, but some animals are more equal than others.'
s = s.rstrip('.').lower()
s_tokens = s.split()
print('There are', len(s_tokens), 'words in the sentence.')
````

## Control Flow

* Programming becomes interesting once we depart from straight-line programs!

* **Control flow** is the order in which statements are executed or evaluated
* In Python, there are three main categories of control flow:
  * **Branches** (conditional statements) – execute only if some condition is met
  * **Loops** (iteration) – execute repeatedly 
  * **Function calls** – execute a set of distant statements and return back to the control flow

## Iteration: `while` with conditional statements

Loops can contain branches. Here, the loop repeats until the guess is right; then `repeat = False` ends it:

````python
correct = 25
repeat = True

while repeat:
    guess = int(input("Guess which number from 1 to 100 I'm thinking of? "))
    
    if guess > correct + 10 or guess < correct - 10:
        print("You are quite far. Try again.")
    elif guess != correct:
        print("You are very close. Try again.")
    else:
        print("That's right!")
        repeat = False
        
````

## `range()`

* In-built function that produces an immutable ordered non-scalar object of type `range`
* Initiate as `range([start], stop, [step])`. If omitted, `start = 0` and `step = 1`. 
* Function produces progression of integers `[start, start + step, start + 2*step, ..., start + i*step]` 
  * If `step > 0`: `start + i*step < stop` 
  * If `step < 0`: `start + i*step > stop` 

````python
print(range(6))
print(list(range(6)))
````

````python
# use list() and range() to create the list [2, 4, 6, 8]
````

## `range()` is essential for `for`-loops

With `range()`, a `for` loop repeats a block of code a given number of times:

````python
for i in range(6):
    print(i, end=' ')
print() 

for i in range(1, 6):
    print(i, end=' ')
print()
    
for i in range(1, 6, 2):
    print(i, end=' ')
    
````

## Indexing Lists with `range(len(L))`

To go through the *positions* of a list instead of its elements, combine `range()` and `len()`:

````python
mylist = ['a', 'b', 'c', 'd']
for i in range(len(mylist)):
     print('index', i, '-', mylist[i])

# This application is so common that there is a function for it
for i, val in enumerate(mylist):
     print('index', i, '-', val) 
````

* This is especially useful when you need to go simultaneously over two different lists of the same length

````python
mylist1 = ['a', 'b', 'c', 'd']
mylist2 = [1, 2, 3, 4]
for i in range(len(mylist1)):
     print(mylist1[i] + str(mylist2[i]), end=', ')
````

# Writing (Simple) Numerical Programs

## Finding Square Roots

````python
# Find an approximation to the square root of a non-negative number

x = 2500  # This is the number we are given

ans = 0  # Start checking from the lowest possible number

# Increment ans until all options exhausted
while ans**2 < x:
    ans += 1
    
if ans**2 != x:
    print(x, 'is not a perfect square')
else:
    print('The square root of', x, 'is', ans) 
    
````

## Exhaustive Enumeration

<div style="width: 55%; margin-right: 28px;">

* Surprisingly useful as computers are quite fast these days!

## Approximation with Exhaustive Enumeration

````python
# Find an approximation to the square root of a non-negative number 
# using exhaustive enumeration

x = 25

epsilon = 0.01  # Precision of approximation  
step = epsilon**2  # Increment in smaller steps than required precision

num_guess = 0  # Keep track of iteration steps
ans = 0

# Increment ans with step until close enough or until all options exhausted
while abs(ans**2 - x) >= epsilon and ans <= x:
    ans += step
    num_guess += 1
    
if abs(ans**2 - x) >= epsilon:
    print('Failed to find close approximation to the square root of', x)
else:
    print('Found', ans, 'to be close approximation to the square root of', x)
    
print('num_guess =', num_guess)
    
````

*In divide and conquer algorithms you divide the problem into smaller pieces until you can solve them, then reassemble the pieces to find the final solution.*

## Approximation with Bisection Search

````python
# Find an approximation to the square root of a non-negative number 
# using bisection search

x = 25

epsilon = 0.01  # Precision of approximation  
num_guess = 0 # Keep track of iteration steps

# Define interval for search
low = 0
high = max(1, x)  # Use 1 if 0 <= x < 1

# Start in the middle
ans = (high + low) / 2 

# Narrow down search interval until ans close enough
while abs(ans**2 - x) >= epsilon:
    # If we are underestimating, shift to upper half of interval
    if ans**2 < x:
        low = ans
    # Otherwise, shift to lower half
    else:
        high = ans
    # For next iteration, start again in the middle of the new interval
    ans = (high + low) / 2
    num_guess += 1
    
print('Found', ans, 'to be close approximation to the square root of', x)
    
print('num_guess =', num_guess)
````

## Newton-Raphson Method for Finding Polynomial Roots

<div style="display: flex;">
<div style="width: 55%; margin-right: 28px;">

* $x^2 - 25$ is a polynomial $p$
* Newton proved a theorem that implies that if $a$ is an approximation to the root of $p=0$, then $a - \frac{p(a)}{p'(a)}$ is a better approximation
* $p'$ is the first derivative of $p$. For $p = x^2 - 25$, $p' = 2x$

## Approximation with Newton-Raphson Method

````python
# Find an approximation to the square root of a number 
# using Newton-Raphson method
# Find x such that x**2 - 25 is within epsilon of 0.01

k = 25

epsilon = 0.01  # Precision of approximation  
num_guess = 0 # Keep track of iteration steps

# Initialize first guess
ans = k

# Use Newton's theorem until ans close enough
while abs(ans**2 - k) >= epsilon:
    ans = ans - ( (ans**2 - k) / (2*ans) )
    print(ans)
    num_guess += 1
    
print('Found', ans, 'to be close approximation to the square root of', k)
    
print('num_guess =', num_guess)
````
