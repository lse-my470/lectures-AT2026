# Scalar Data Types and Operators

## From Last Week: Objects, Data Types, and Expressions

* Computer programs manipulate objects
* Objects have types
  * Scalar — indivisible
  * Non-scalar — with internal structure
* Expressions combine objects and operators

## Scalar Data Types

**Scalar** objects are indivisible

* `int` → used to represent integer (e.g., $-5$, $8$, $1894$)
* `float` → Used to represent real numbers (e.g., $5.0$, $14.7$, $-23.92$)
* `bool` → Used to represent the Boolean values `True` and `False`
* `NoneType`
* (String is non-scalar in Python)


````python
print(type(2))
print(type(1.125))
print(type(True))
print(type(None))
print(type('a'))
````

## Converting between Scalar Data Types

* Use the name of a type to convert values to that type

````python
a = float(123)
b = int('32')
print(a, b)
````

## Operators

**Operators** combine objects into expressions

* Arithmetic → calculate with numbers (e.g., `7 / 2`, `2 + 3`)
* Boolean → combine truth values (e.g., `True and False`)
* Comparison → compare two objects; the result is a `bool` (e.g., `3 > 2`)
* Assignment → bind an object to a variable name (e.g., `x = 3`)

## Arithmetic Operators

**Arithmetic operators** calculate with numbers

| Operator | Meaning |
|:--|:--|
| `+` | addition |
| `-` | subtraction |
| `*` | multiplication |
| `/` | division |
| `%` | modulus: remainder of a division (`7 % 2` is `1`) |
| `//` | floor division: division rounded down to a whole number (`7 // 2` is `3`) |
| `**` | exponent |

````python
print(2 + 2, 3 * 2, 7 / 2)
print(7 % 2, 7 // 2, 2 ** 3)
````

````python
# use % to check whether 17 is an even or an odd number
````

## Operators Depend on the Object Type

The same operator can do different things, depending on the types of the objects it is applied to

| Operator | Numbers | Strings |
|:--|:--|:--|
| `+` | addition: `2 + 2` is `4` | concatenation: `'a' + 'bc'` is `'abc'` |
| `*` | multiplication: `3 * 2` is `6` | repetition: `3 * 'a'` is `'aaa'` |

````python
# + and * have different meanings depending on the types of objects with which they are used
print(2 + 2)
print('a' + 'bc')
print(3 * 2)
print(3 * 'a' + 'h!')
````

## Comparison Operators

**Comparison operators** compare two objects; the result is a `bool` (i.e., `True` or `False`)

| Operator | Meaning |
|:--|:--|
| `==` | equals |
| `!=` | does not equal |
| `>` | is greater than |
| `<=` | is less than or equal, etc. |

````python
print(3 == 3, 3 != 4)
print(5 > 2, 5 <= 4)
print('a' == 'A')  # comparison of strings is case-sensitive
````

````python
# check whether 2 * 'a' is equal to 'aa'
````

## Boolean Operators

**Boolean operators** combine truth values; the result is a `bool`

| Operator | Meaning |
|:--|:--|
| `and` | `True` if both sides are `True` |
| `or` | `True` if at least one side is `True` |
| `not` | reverses the value: `not True` is `False` |

````python
print(True and False)
print(True or False)
print(not False)
````

## Assignment Operators

**Assignment operators** bind an object to a variable name

| Operator | Meaning |
|:--|:--|
| `=` | assign right operand to left operand |
| `+=` | add right operand to left operand and assign to left operand |
| `-=` | subtract right operand from left operand and assign to left operand, etc. |

````python
a = 2
a += 3  # Equivalent to a = a + 3
print(a)
````

````python
# start with b = 10, then use *= to double b and print it
````

## Comparison vs. Assignment Operators

A single `=` **assigns** an object to a name; a double `==` **compares** two objects and returns a `bool`

````python
a = 2  # This is assignment
print(a == 1)  # This is test for equality. It returns bool.
````

## Non-Scalar Data Types

**Non-scalar** objects have internal structure

* `list` → a mutable ordered sequence of values (e.g., `[1, 2, 'a']`)
* `tuple` → an immutable ordered sequence of values (e.g., `(1, 2, 'a')`)
* `str` (string) → an immutable ordered sequence of characters (e.g., `'abc'`)
* `set` → a mutable unordered collection of unique values (e.g., `{1, 2, 'a'}`)
* `dict` (dictionary) → a set of key/value pairs (e.g., `{1: 'a', 2: 'b'}`)

## String Methods: Formatting

* `S.upper()` → change to upper case
* `S.lower()` → change to lower case
* `S.capitalize()` → capitalize the first word
* `S.find(S1)` → return the index of the first instance of input

````python
print('Make me scream!'.upper())
x = 'make this into a proper sentence'
print(x.capitalize() + '.')

print('Find the first "i" in this sentence.'.find('i'))
````

## String Methods: `strip` and `replace`

* `S.replace(S1, S2)` → find all instances of S1 and change to S2
* `S.strip(S1)` → remove whitespace characters from the beginning and end of a string (useful when reading in from a file)

````python
x = ' This is a long sentence that we will use as an example.\n'
print(x.replace('s', 'S'))
print(x.strip())
print(x.replace(' ', ''))
````

## String Methods: `split` and `join`

* `S.split(S1)` → split the string into a list
* `S.join(L)` → combine the input sequence into a single string

````python
x = 'this is a collection of words i would like to break it into tokens'
y = x.split()    # default is to split on ' '
print(y)
print(x.split('o')) 

x_new = '-'.join(y)
print(x_new)
````

````python
# use split() and join() to turn 'a b c' into 'a+b+c'
````

## Strings Are Immutable

A string can never be changed: string methods **return a new string** and leave the original as it is. To keep working with the result, **store it in a variable**.

````python
x = 'hello world'
print(x.upper())   # returns a new string ...
print(x)           # ... but x itself is unchanged
````

````python
x_upper = x.upper()   # store the result in a new variable
print(x_upper)
print(x)
````

````python
x = x.capitalize()    # or overwrite the old variable
print(x)
````

## Unordered Types vs. Sequences

**Order** determines how you can access the elements of a non-scalar object

* Unordered types: `set`, `dict` → elements have no position; check membership with `in`, or look up values by key
* Ordered (sequence) types: `str`, `list`, `tuple` → every element has a position, so you can access it by its index (e.g., `'abc'[0]`)

````python
st = {1, 2, 2, 'a', 'b'}  # set: no order, no duplicates
print(st)
print('a' in st)  # check membership with in
ls = [1, 2, 2, 'a', 'b']  # list: order and duplicates kept
print(ls)
print(ls[3])  # access an element by its index
````

## Set Methods

**Set methods** combine two sets, as in set theory: write them as methods or operators

| Method | Infix Operator | Elements … |
|:--|:--|:--|
| `S1.union(S2)` | `S1 \| S2` | in S1 or S2, or both |
| `S1.intersection(S2)` | `S1 & S2` | in both S1 and S2 |
| `S1.difference(S2)` | `S1 - S2` | in S1 but not in S2 |
| `S1.symmetric_difference(S2)` | `S1 ^ S2` | in S1 or S2 but not both |

## Sequence Operations: Slicing

* Use `elem[start:end]` to get sub-sequence starting from index `start` and ending at index `end-1`

````python
ls = [10, 20, 30, 40, 50]
print(ls[1:4]) 
print(ls[:3])
print(ls[1:])

ls[:] == ls[0:len(ls)]
````

## Sequence Operations: Extended Slices

* Use `elem[start:end:step]` to get sub-sequence starting from index `start`, in steps of `step`, ending at index `end-1`

````python
ls = [10, 20, 30, 40, 50]
print(ls[::2])  # get elements with even indices
print(ls[::-1])  # get elements in reverse order
````

## Mutability

* Immutable types: `str`, `tuple`, and all scalars
* Mutable types: `list`, `set`, `dict`

**Objects of mutable types can be modified once they are created.**

````python
dic = {1:'a', 2:'b'}
dic[3] = 'c'
print(dic)

ls = [5, 4, 1, 3, 2]
ls.sort()
print(ls)
````

## Mutability Can Be Quite Convenient

There are several useful list methods, see the [documentation of mutable sequence types](http://docs.python.org/3/library/stdtypes.html#mutable-sequence-types):

| Method | What it does |
|:--|:--|
| `L.append(e)` | adds element `e` to the end of `L` |
| `L.extend(L1)` | adds all elements of list `L1` to the end of `L` |
| `L.insert(i, e)` | inserts element `e` at index `i` |
| `L.remove(e)` | removes the first occurrence of `e` |
| `L.pop(i)` | removes the element at index `i` and returns it (the last one if `i` is left out) |
| `L.sort()` | sorts `L` in place |
| `L.reverse()` | reverses the order of `L` in place |

````python
ls1 = [1, 2, 3]
ls1.append(4)
print(ls1)
ls1.extend([5, 6, 7, 8, 9, 10])
print(ls1)
````

````python
# use insert() to put 0 at the beginning of ls1
````

## List Methods: `L.sort()` vs. `sorted(L)`

````python
mylist = [4, 5, 2, 1, 3]
mylist.sort()  # Sorts in-place. It is more efficient but overwrites the input.
print(mylist)

mylist = [10, 9, 6, 8, 7]
sorted(mylist) 
print(mylist)

newlist = sorted(mylist)  # Creates a new list that is sorted, not changing the original.
print(mylist, newlist)
````
