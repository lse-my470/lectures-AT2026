## Recap: Testing and Debugging

* **Testing:** running a program to confirm that it works as intended
* **Debugging:** fixing a program that does not work as intended

Design your programs to make both easier: break them into components (**modularity**), and document constraints and expectations in the function specifications (**docstrings**).

## Recap: Unit Testing

A unit test checks that a small unit of the code works as expected, e.g. a function or a method of a class. How do we choose the test cases?

* **Random testing:** explore a large set of random input values
* **Black-box testing:** focus on the **specification**. Test boundary conditions (e.g. an empty list), all natural partitions (e.g. positive and negative numbers), and extremes (e.g. very large values)
* **Glass-box testing:** focus on the **code**. Test every path through the program: both branches of `if` statements, `except` clauses, `for` and `while` loops (including not entering the loop), and recursive calls

**Combine black-box and glass-box testing!**

## Recap: Steps of Testing

1. **Make sure the code runs:** remove syntax and static semantic errors
2. **Unit testing:** test individual functions and methods
3. **Regression testing:** fix the bugs and perform step 2 again
4. **Integration testing:** test the program as a whole

**If a test fails:** fix the bug in the code, or rewrite the function specification.

## Structuring Your Programs with `.py` Files

* Keep functions and classes that belong together in the same `.py` file, e.g. `tools.py`, `plots.py`, `test_tools.py`
* Keep the code that runs the program (the main execution code) in a separate `.ipynb` or `.py` file
* Import the modules to use their functions and classes:

```python
import tools               # tools.py in the same folder
tools.running_sum([1, 2, 3])
```

## General Structure of `.py` Files

```python
# Start with importing the modules you need
import some_module

# Then define global variables (constants), if needed
GLOBAL_VAR = None

# main() says what to do if the file is run as a script
def main():
    pass

# Define classes and functions
def your_function():
    pass

# End the file with this statement
if __name__ == '__main__':
    main()
```

## Understanding `if __name__ == '__main__':`

* Whenever Python runs a file, it sets some special variables. `__name__` is one of them
* If we **run** the file directly, `__name__` is `'__main__'`
* If we **import** the file, `__name__` is the name of the file without `.py`, e.g. `'some_module'`

* So the code under `if __name__ == '__main__':` runs only when you run the file itself, not when you import it
* This lets a file work both as a script and as a module

# Unit Testing

## Testing with `unittest`

* Put the code in one module and the tests in another, e.g. `leading.py` and `test_leading.py`
* Write one test class (a subclass of `unittest.TestCase`) for each function or method you test
* Each test is a method whose name starts with `test`. Check the result with an assert method, e.g. `assertEqual(expected, actual, message)` (see the [documentation](https://docs.python.org/3/library/unittest.html#assert-methods))
* End the test file with `unittest.main()` under `if __name__ == '__main__':`; do not write your own `main()`
* Keep the tests out of the main program
* Run the tests in the terminal: `python -m unittest -v test_leading`

The function `leading_substrings()` below returns all substrings that start at the beginning of a string.

4. Run the tests in the terminal or with the second cell below. All tests should pass (`OK`).
5. Break the function on purpose, e.g. with `range(len(s) - 1)`, and run the tests again: does one of your tests fail?

````python
def leading_substrings(s):
    """Takes string s as input and returns a list of all
    the substrings that start from the beginning.
    E.g., leading_substrings('bear') will return
    ['b', 'be', 'bea', 'bear'].
    """
    return [s[:i+1] for i in range(len(s))]
````

## Exercise 4

Rewrite the function below. Instead of checking beforehand whether the dictionary has a list with the name `lname`, **try** to get the list:

1. `try`: get the list `dic[lname]`.
2. `except`: if there is no such list (catch only this exception), create an empty list and print `Added <lname> to dictionary`.
3. `else`: print `<lname> already has <n> items`.
4. `finally`: append the element and print `Item <element> successfully added to <lname>`.

````python
def add_to_list_in_dict(dic, lname, element):
    """Adds element to the list with the name lname in dictionary dic."""
    if lname not in dic:
        dic[lname] = []
        print('Added', lname, 'to dictionary')
    else:
        x = dic[lname]
        print(lname, 'already has', len(x), 'items')

    dic[lname].append(element)
    print('Item', element, 'successfully added to', lname)

# Test
dic = {'a': [], 'b': [1]}
add_to_list_in_dict(dic, 'b', 4)
add_to_list_in_dict(dic, 'c', 1)
print(dic)

# Expected output:
# b already has 1 items
# Item 4 successfully added to b
# Added c to dictionary
# Item 1 successfully added to c
# {'a': [], 'b': [1, 4], 'c': [1]}
````

## Exercise 5

The function below computes a party's vote share. Its docstring states what it assumes, but nothing checks it.

Add an `assert` statement with a short message for each assumption in the docstring.

````python
def vote_share(votes, total):
    """Assumes votes and total are integers, total > 0, and
    0 <= votes <= total. Returns votes as a percentage of total,
    rounded to one decimal.
    """
    return round(votes / total * 100, 1)

# Test
tests = [(40, 120), (35, 0), (130, 120), (40.5, 120)]
for votes, total in tests:
    try:
        print(vote_share(votes, total))
    except AssertionError as e:
        print('AssertionError:', e)

# Expected output: 33.3, then three AssertionErrors with your messages
````
