## Glass-Box Testing

<div class="fragment" style="display: flex; margin-top: 0px; font-family: system-ui; line-height: 1.35; color: #212121;"><div style="width: 16px; min-width: 16px; height: 16px; margin: 9px 20px 0 0; background-color: #b0603a;"></div><div>Focus on the code: easier to construct and make more thorough</div></div>
<div class="fragment" style="display: flex; margin-top: 12px; font-family: system-ui; line-height: 1.35; color: #212121;"><div style="width: 16px; min-width: 16px; height: 16px; margin: 9px 20px 0 0; background-color: #b0603a;"></div><div>Aim for <b>path-complete</b> tests: test every possible path through the program</div></div>
<pre class="fragment" style="margin-top: 28px; line-height: 1.2; letter-spacing: 0;"><code style="line-height: 1.2; letter-spacing: 0;">                  ┌──────────────────────────────────────────┐
                  │                    ┌─ True ──► line 3 ─┐ │
   test input ──► │ line 1 ──► line 2 ─┤                   ├─┼──► output
                  │                    └─ False ─► line 5 ─┘ │
                  └──────────────────────────────────────────┘</code></pre>

<div style="margin-top: 22px;"></div>

<ul>
<li class="fragment">Test both branches of <code>if</code> statements</li>
<li class="fragment">Test all <code>except</code> clauses</li>
<li class="fragment">For <code>for</code> loops, test not entering the loop, entering once, and entering multiple times</li>
<li class="fragment">For <code>while</code> loops, test the same, and exiting the loop in every possible way</li>
<li class="fragment">For recursive functions, test no recursive calls, one recursive call, and multiple recursive calls</li>
</ul>

## Example: Glass-Box Testing

````python
def absolute(x):
    """Assumes x is an int.
    Returns x if x >= 0 and -x otherwise.
    """
    if x < -1:
        return -x
    else:
        return x

print(absolute(2), absolute(-2))  # a path-complete test: both branches
````

````python
# predict, then check: what does absolute(-1) return?
````

* Path-complete test: `2` and `-2`
* But `absolute(-1)` returns `-1`!
* **Always combine with black-box testing!**

## Example: Unit Testing

````python
# Example taken from Gries et al. 2013. Practical Programming: An Introduction
# to Computer Science Using Python 3

def running_sum(ls):
    """Modifies list ls so that it contains the running sums of its original items.
    E.g., running_sum([1, 2, 3]) updates the list to [1, 3, 6].
    """
    for i in range(len(ls)):
        ls[i] = ls[i - 1] + ls[i]
````

````python
# test running_sum() on the list [1, 2, 3]: does it work?
````

## Which Cases Should We Test?

| Case   | List before  | List after
| :----: |:------:| :----------------------
| Empty list | `[]`         | `[]`
| One-item list | `[2]`    | `[2]`
| Two-item list | `[2, 5]`    | `[2, 7]`
| Multiple items, all negative | `[-1, -5, -3, -4]`    | `[-1, -6, -9, -13]`
| Multiple items, all positive | `[4, 2, 3, 6]`    | `[4, 6, 9, 15]`
| Multiple items, all zero | `[0, 0, 0, 0]`  | `[0, 0, 0, 0]`
| Multiple items, mixed | `[4, 0, 2, -5]`    | `[4, 4, 6, 1]`

## Unit Testing with `unittest`

* [`unittest`](https://docs.python.org/3/library/unittest.html) is part of Python's standard library
* `running_sum()` is in the file `tools.py`, the tests are in `test_tools.py`
* Each test is a method of a subclass of `unittest.TestCase`, and its name starts with `test`

```python
import unittest
import tools

class TestRunningSum(unittest.TestCase):
    """Tests for running_sum."""

    def test_running_sum_empty(self):
        """Test an empty list."""
```

## Handling All Exceptions with `try` and `except`

````python
def div(a, b):
    """Prints the result from dividing a by b."""
    try:
        print(a / b)
    except:
        print("Cannot divide given values.")

    print("We are learning about try/except")

div(1, "2")
````

A bare `except` catches **every** exception, including the ones caused by our own bugs.

## Handling Specific Exceptions with `try` and `except`

* If you know that a line of code may raise an exception when executed, handle it!

````python
def div(a, b):
    """Returns the result from dividing a by b."""
    try:
        return a / b
    except ZeroDivisionError:
        print("Division by zero is undefined.")
    except TypeError:
        print("Function should be used with numerical types.")

print(div(1, 0))
````

## `try`, `except`, `else`

* `else`: runs only if **no** exception has been raised

````python
def div_add(a, b, c):
    """Returns a/b + c."""
    try:
        x = a / b
    except ZeroDivisionError:
        print('Cannot divide', a, 'by', b)
    else:
        return x + c

print(div_add(1, 2, 3))
````

## `try`, `except`, `finally`

* `finally`: runs **always**, regardless of whether an exception has been raised or handled and `return` called
* Good for clean-up, e.g. closing a file

````python
def div(a, b):
    """Returns the result from dividing a by b."""
    try:
        return a / b
    except ZeroDivisionError:
        print("Division by zero is undefined.")
    finally:
        print("This will always print.")

print(div(4, 2))
````

## Raising Exceptions with `raise`

**Instead of failing silently and second-guessing wrong inputs, alert the user that something has gone wrong and stop execution!**

```python
raise ExceptionName(arguments)
```

* `ExceptionName` can be a built-in exception such as `ValueError`, or a custom exception
* `arguments` is typically a string that describes why the exception is raised

````python
def sqrt(x):
    """Assumes x is a number >= 0. Returns the square root of x."""
    if x < 0:
        raise ValueError('x must be >= 0, got ' + str(x))
    return x ** 0.5
````
