## Time Complexity: Example 1

`sum_of_n(n)` computes $1 + 2 + \ldots + n$ with a loop:

```python
def sum_of_n(n):
    """Assumes n is a positive int. Returns the sum of 1 to n."""
    res = 0
    for i in range(1, n + 1):
        res += i
    return res
```

How should we think about the question "How long does this program take to run?"

**Benchmarking:** run the program and measure the time it takes

```python
import time
for n in [100000, 1000000, 10000000]:
    start = time.time()
    res = sum_of_n(n)
    end = time.time()
    print(n, end - start)
```

## Limitations to Benchmarking

Running time depends on:

  * The speed of the computer
  * The efficiency of programming language implementation (in our case, Python) on the computer
  * The value of the input

## Steps Depend on the Input

<div style="box-sizing: border-box; padding: 12px 18px; border-radius: 8px; background-color: #e3ede4; font-family: system-ui; line-height: 1.35; color: #212121;">Express measure as <b>function of the size of inputs</b></div>

```python
def total(L):
    """L is a list of numbers"""
    res = 0
    for x in L:
        res += x
    return res

print(total([4, 8, 15]))
print(total(list(range(1000))))
```

## Upper Bound: The Worst Case

<div style="box-sizing: border-box; padding: 12px 18px; border-radius: 8px; background-color: #f7efe6; font-family: system-ui; line-height: 1.35; color: #212121;">For some functions, there is <b>no 1:1 mapping</b> from input size to number of steps: inputs of the same size $n$ can take different numbers of steps</div>

```python
def linear_search(L, x):
    """Assumes L is a list and x is any value.
    Returns True if x is in L, False otherwise."""
    for e in L:
        if e == x:
            return True
    return False

print(linear_search([3, 8, 5, 1], 3))   # 3 is the first element
print(linear_search([3, 8, 5, 1], 7))   # 7 is not in the list
```

# Order of Growth

## Approximation for Large Inputs

<div style="box-sizing: border-box; padding: 12px 18px; border-radius: 8px; background-color: #e3ede4; font-family: system-ui; line-height: 1.35; color: #212121;">Use an <b>approximation</b> as the size of the input gets extremely large</div>

```python
for n in [10, 1000, 1000000]:
    steps = 3 * n + 2                # T(n) for total(L)
    print(n, steps, 3 * n / steps)   # share of the 3n term
```

# Recognising Complexity in Code

## Important Computational Complexity Classes

| Big-O notation | Running time | Pattern in code |
| :--- | :--- | :--- |
| $O(1)$ | constant | no loop or recursion whose number of steps depends on $n$ |
| $O(\log n)$ | logarithmic | the problem shrinks by a factor in each step |
| $O(n)$ | linear | one pass over the input; $n$ recursive calls |
| $O(n \log n)$ | log-linear | sorting |
| $O(n^c)$ | polynomial | $c$ loops over the input, nested inside each other |
| $O(c^n)$ | exponential | each call branches into several recursive calls; all combinations |

## Comparison of Complexity Classes

## Constant Complexity: $O(1)$

* Running time is independent of the input size

## Constant Complexity: Example

`sum_of_n` from Part 1, now with a formula instead of a loop: $1 + 2 + \ldots + n = n(n+1)/2$

```python
def sum_of_n(n):
    """Assume n is a positive integer.
    Calculate the sum of all integers from 1 to n, inclusive.
    """
    return n * (n + 1) // 2

for i in [100000, 1000000, 10000000]:
    start = time.time()
    res = sum_of_n(i)
    end = time.time()
    print(res, end - start)
```

## Logarithmic Complexity: $O(\log n)$

* Running time grows as the log of at least one of the inputs
* Typically, we use base 2 but the base of the logarithm does not matter as $\log_2 x = \log_2 10 \cdot \log_{10} x$ and we ignore constants
* If we use base 2, then we can say: **Every time the size of the input doubles, the algorithm performs one additional step.**
* **How to spot it:** the problem shrinks by a constant factor in each step, e.g. the loop variable is halved (`n = n // 2`) instead of decreased by one (`n = n - 1`, which is linear)

## Logarithmic Complexity: Bisection Search

From week 3: each guess halves the search interval

```python
def bisec_search(x, epsilon):
    """Assumes x >= 0. Returns an approximation within epsilon
    to the square root of x, and the number of guesses needed."""
    low, high = 0, max(1, x)    # define interval for search
    guess = (low + high) / 2    # start in the middle
    num_guess = 1
    while abs(guess**2 - x) >= epsilon:
        if guess**2 < x:
            low = guess
        else:
            high = guess
        guess = (low + high) / 2
        num_guess += 1
    return guess, num_guess

bisec_search(25, 0.01)
```

## Linear Complexity: $O(n)$

* **One pass over the input**: each element is touched a constant number of times
* Also linear: **$n$ recursive calls**, each doing a constant amount of work

## Linear Complexity: Examples

One pass over the input:

```python
n = 10
for x in range(n):     # n iterations
    print(x, end=' ')
```

$n$ recursive calls: `factorial(n)` computes $n! = n \cdot (n-1) \cdot \ldots \cdot 2 \cdot 1$ (week 4)

```python
def factorial(n):
    """Assumes n is a positive integer. Returns n!."""
    if n == 1:
        return 1
    else:
        return n * factorial(n - 1)

factorial(5)
```

## Log-Linear Complexity: $O(n \log n)$

More complicated than previous complexity classes and we won't go into detail. **But:**

* Many practical algorithms are log-linear
* For example, **sorting a list**
* **How to spot it:** a logarithmic step repeated for each of the $n$ elements ($n \cdot \log n$), or a call to `sorted()` or `.sort()`
* Next week, we'll study different sorting algorithms

## Log-Linear Complexity: Example

Sorting a list of $n$ random numbers:

```python
import random

for n in [500000, 1000000, 2000000, 4000000]:
    L = [random.random() for i in range(n)]
    start = time.time()
    L.sort()
    end = time.time()
    print(n, round(end - start, 3))
```

* When $n$ doubles, sorting takes **a bit more than twice** as long: $n \log n$ grows slightly faster than $n$

## Polynomial Complexity: $O(n^c)$

* E.g. quadratic $O(n^2)$ or cubic $O(n^3)$
* **How to spot it:** loops over the input nested inside each other, each running up to $n$ times: two levels give $O(n^2)$, three give $O(n^3)$

## Polynomial Complexity: Example

Two nested loops over the input, $O(n^2)$:

```python
n = 10
test = 0
for i in range(1, n + 1):          # n iterations
    for j in range(1, n + 1):      # n iterations each
        test += i * j
print(test)
```

## Exponential Complexity: $O(c^n)$

* Many important problems are inherently exponential
* Exponential algorithms are considered too slow to be practical
* In such cases, we can find algorithms that provide approximate solutions or at least address specific instances of the problem
* **How to spot it:** several recursive calls per call, or trying all combinations

## Exponential Complexity: Example

**Fibonacci numbers:** each number is the sum of the two before it, $F_n = F_{n-1} + F_{n-2}$, starting with $F_0 = 0$ and $F_1 = 1$: 0, 1, 1, 2, 3, 5, 8, 13, ...

Two recursive calls per call:

```python
def fib(n):
    """Assumes n >= 0 is an int. Returns the n-th Fibonacci number."""
    if n <= 1:
        return n
    else:
        return fib(n - 1) + fib(n - 2)   # two recursive calls

print(fib(10))
```

* Each call makes **two** more calls, so the number of calls roughly **doubles at each level**: $1 + 2 + 4 + \ldots + 2^{n-1} \approx 2^n$, i.e. $O(2^n)$. `fib(10)` needs 177 calls, `fib(30)` almost 2.7 million

## Hidden Loops in Practice: Growing a List

Both functions build the same list. Knowing what each operation costs makes one of them much faster.

* `L.append(k)` is $O(1)$: changes the list **in place**; existing elements are not touched
* `L + [k]` is $O(l)$: builds a **new** list, copying all $l$ entries of `L` one by one

<div class="code-columns"></div>

*The method `append()` adds an item in place and since it adds to the end of the list, it does not have to touch the existing items*. In contrast, the list concatenation operation touches every item of the first list since it creates a copy of it.

*Trace of `l_concatenate(5)`: `[]` → `[0]` → `[0, 1]` → `[0, 1, 2]` → … In iteration `i`, `lst` already has `i` elements, and all of them are copied into the new list, so this one line costs about `i` steps. In total 0 + 1 + … + (k − 1) ≈ k²/2 copies.*

*What is copied are the references to the objects, one per element; the objects themselves (here the integers) are not duplicated (shallow copy, as with `L[:]` in week 2). Each copy is fast, but there is one per element, so the cost grows with the length of the list.*

```python
def l_concatenate(k):
    lst = []
    for i in range(k):
        lst = lst + [i]   # copies i entries
    return lst

l_concatenate(4)
```

```python
def l_append(k):
    lst = []
    for i in range(k):
        lst.append(i)     # adds 1 entry
    return lst

l_append(4)
```

* In the loop, the copy gets longer every time: `l_concatenate(k)` is $O(k^2)$, `l_append(k)` is $O(k)$

## Python Lists: Timing Both Versions

Predict first: which version is faster, and by how much?

```python
def timit(fnc, func_arg):  # functions can be passed as arguments
    start = time.time()
    fnc(*func_arg)  # Pass items from iterable as function arguments
    end = time.time()
    print(fnc.__name__, '\t', end - start)

timit(l_concatenate, (10000,))
timit(l_append, (10000,))
```

## Time Complexity of Python List Methods

$O(1)$

* Indexing `[]`, `append()`, `pop()`

$O(n)$

* Iteration, `pop(i)`, `insert()`, `del`, `remove()`, `in`, `count()`, `reverse()`, `min()`, `max()`

$O(n \log n)$

* `sort()`

## Time Complexity of Python Dictionary Methods

$O(1)$ on average

* Indexing `d[k]`, assignment `d[k] = v`, `in`, `del`, `len()`

$O(n)$

* Iteration (over the keys, `values()`, `items()`)

Sets work the same way: `in` takes $O(1)$ on average. For a list, `in` takes $O(n)$.

## Complexity Analysis in Practice

* There is no fixed list of possible runtimes, e.g. $O(\sqrt n)$, $O(n!)$, ...
* Runtimes can be a function of more than one input, e.g. $O(nk)$, $O(n + k)$

* Big-O describes the rate of increase in runtime, not the actual runtime
    * Benchmarking is still a useful tool

* Big-O ignores constants but they often matter in practice
    * If you reduce your code from three loops to one loop, Big-O will not change but the difference may be on the order of hours or days!

* There is often a tradeoff between time and space efficiency
* **You need a balance between runtime, space, implementation time, maintainability, and readability**

## Space Complexity: Examples

* $O(1)$: besides the input, only the loop variable `i` is stored, however large $n$ is

```python
def print_n_times(s, n):
    for i in range(n):
        print(s)
```

* $O(n)$: the new list `ls` has $n$ entries, so the memory grows with $n$

```python
def combine_n_times(s, n):
    ls = [s for i in range(n)]
    return ls
```
