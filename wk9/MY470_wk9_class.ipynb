## Recap: Order of Growth

* We count **steps** (not seconds) as a function of the input size $n$, in the **worst case**
* Big-O keeps the **fastest-growing term** and drops constants: $3n^2 + 5n + 100$ is $O(n^2)$
* Where to look:
    * **Loops:** how often does each loop run? Nested loops multiply
    * **Function and method calls:** what do they cost? For a list, `x in L` and `L.count(x)` are $O(n)$
    * **Recursive calls:** how many are there?

## Recap: How to Spot the Complexity Class

| Big-O notation | Running time | Pattern in code |
| :--- | :--- | :--- |
| $O(1)$ | constant | no loop or recursion whose number of steps depends on $n$ |
| $O(\log n)$ | logarithmic | the problem shrinks by a factor in each step |
| $O(n)$ | linear | one pass over the input; $n$ recursive calls |
| $O(n \log n)$ | log-linear | sorting |
| $O(n^c)$ | polynomial | $c$ loops over the input, nested inside each other |
| $O(c^n)$ | exponential | several recursive calls per call; all combinations |

# Measuring Running Time

## Timing Code

`time.perf_counter()` works like `time.time()` from the lecture, but is more precise for short durations. Running times vary from run to run, so `time_it()` runs the code three times and keeps the shortest time. You need it for the rest of the lab.

```python
import time

def time_it(func, arg, repeat=3):
    """Calls func(arg) repeat times. Returns the shortest running time
    in seconds, i.e. the run that was disturbed least."""
    times = []
    for r in range(repeat):
        start = time.perf_counter()
        func(arg)
        times.append(time.perf_counter() - start)
    return min(times)

numbers = list(range(1000000))
print(time_it(sum, numbers))
```

```python
def first_and_last(L):
    """Returns the sum of the first and the last element of list L."""
    return L[0] + L[-1]

def count_last(L):
    """Returns how often the last element of list L occurs in L."""
    return L.count(L[-1])

def count_equal_pairs(L):
    """Returns the number of pairs of positions i < j with L[i] == L[j]."""
    count = 0
    for i in range(len(L)):
        for j in range(i + 1, len(L)):
            if L[i] == L[j]:
                count += 1
    return count
```

# Choosing Data Structures

## `in`: Lists versus Sets

* For a **list**, `x in L` checks the elements one by one: $O(n)$
* For a **set** (or the keys of a dictionary), Python finds `x` directly, without looking at the other elements: $O(1)$ on average

Run the cell. We search for `-1`, which is not in the data: the worst case for the list.

```python
def search(data):
    """Returns True if -1 is in data."""
    return -1 in data

for n in [100000, 1000000]:
    L = list(range(n))
    S = set(L)
    t_list = round(time_it(search, L) * 1000, 4)   # in milliseconds
    t_set = round(time_it(search, S) * 1000, 4)
    print('n =', n, '| list:', t_list, 'ms | set:', t_set, 'ms')
```
