## Linear Search

<div style="display: flex; margin-top: 0px; font-family: system-ui; line-height: 1.35; color: #212121;"><div style="width: 16px; min-width: 16px; height: 16px; margin: 9px 20px 0 0; background-color: #b0603a;"></div><div>Visit each item in order until you find the item or run out of items (as in week 9)</div></div>

```python
def linear_search(L, x):
    """Assumes L is a list and x is any value.
    Returns True if x is in L, False otherwise."""
    for e in L:
        if e == x:
            return True
    return False

print(linear_search([3, 8, 5, 1], 3), linear_search([3, 8, 5, 1], 7))
```

## Binary Search with a Loop

Like bisection search (week 3): <code>low</code> and <code>high</code> mark the part of the list that is left

```python
def binary_search(L, x):
    """Assumes L is a list with its elements in ascending order.
    Returns True if x is in L, False otherwise."""
    low, high = 0, len(L) - 1
    while low <= high:
        mid = (low + high) // 2
        if L[mid] == x:
            return True
        elif L[mid] > x:
            high = mid - 1      # search the left half
        else:
            low = mid + 1       # search the right half
    return False                # no items left

L = [2, 5, 8, 12, 16, 23, 38, 42, 51, 56, 64, 72, 80, 91, 99]
print(binary_search(L, 23), binary_search(L, 24))
```

## Binary Search with Recursion

From week 4: a function that calls itself. Each call searches one half of the list

```python
def binary_search_rec(L, x):
    """Same as binary_search, but with recursion."""
    # a "wrapper function": it hides the extra parameters low and high
    return b_search(L, x, 0, len(L) - 1)

def b_search(L, x, low, high):
    if low > high:              # base case: no items left
        return False
    mid = (low + high) // 2
    if L[mid] == x:             # base case: found
        return True
    elif L[mid] > x:
        return b_search(L, x, low, mid - 1)     # search the left half
    else:
        return b_search(L, x, mid + 1, high)    # search the right half

print(binary_search_rec(L, 23), binary_search_rec(L, 24))
```

## Linear versus Binary Search: Timing

The worst case for linear search: the item we look for is the last one

```python
import time

L = list(range(10000000))      # sorted: 0, 1, ..., 9999999

start = time.time()
linear_search(L, 9999999)
print('linear search:', round((time.time() - start) * 1000, 3), 'ms')

start = time.time()
binary_search(L, 9999999)
print('binary search:', round((time.time() - start) * 1000, 3), 'ms')
```

## Bubble Sort: Code

```python
def bubble_sort(L):
    """Assumes L is a list of elements that can be compared using >.
    Sorts L in ascending order."""
    # passnum: comparisons in this pass (n - 1, n - 2, ..., 1)
    for passnum in range(len(L) - 1, 0, -1):
        for i in range(passnum):
            if L[i] > L[i + 1]:
                L[i], L[i + 1] = L[i + 1], L[i]     # swap the neighbours

L = [3, 5, 4, 1, 2]
bubble_sort(L)
print(L)
```

## Selection Sort: Code

```python
def selection_sort(L):
    """Assumes L is a list of elements that can be compared using >.
    Sorts L in ascending order."""
    for pos in range(len(L) - 1, 0, -1):    # from the back
        max_pos = 0
        for i in range(1, pos + 1):         # largest item up to pos
            if L[i] > L[max_pos]:
                max_pos = i
        L[pos], L[max_pos] = L[max_pos], L[pos]   # one swap per pass

L = [3, 5, 4, 1, 2]
selection_sort(L)
print(L)
```

## Insertion Sort: Code

```python
def insertion_sort(L):
    """Assumes L is a list of elements that can be compared using >.
    Sorts L in ascending order."""
    for i in range(1, len(L)):
        current = L[i]
        pos = i
        while pos > 0 and L[pos - 1] > current:
            L[pos] = L[pos - 1]     # shift the larger item to the right
            pos -= 1
        L[pos] = current            # insert the current item

L = [3, 5, 4, 1, 2]
insertion_sort(L)
print(L)
```

<div style="display: flex; margin-top: 10px; font-family: system-ui; line-height: 1.35; color: #212121;"><div style="width: 16px; min-width: 16px; height: 16px; margin: 9px 20px 0 0; background-color: #b0603a;"></div><div>Each step moves one item into the result: merging lists with $n$ items in total takes $n$ steps, so $O(n)$</div></div>

## Merging Two Sorted Lists

```python
def merge(left, right):
    """Assumes left and right are sorted lists.
    Returns a new sorted list with the elements of left and right."""
    result = []
    i, j = 0, 0
    while i < len(left) and j < len(right):   # compare the first items
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    # one list is empty now: add the remaining items of the other one
    return result + left[i:] + right[j:]

print(merge([1, 3, 6, 8], [2, 4, 5, 7]))
```

## Merge Sort: Code

Recursion (week 4): <code>merge_sort</code> calls itself on both halves

```python
def merge_sort(L):
    """Assumes L is a list of elements that can be compared using <.
    Returns a new sorted list with the same elements as L."""
    if len(L) <= 1:                 # base case: 0 or 1 items
        return L[:]
    else:
        middle = len(L) // 2
        left = merge_sort(L[:middle])
        right = merge_sort(L[middle:])
        return merge(left, right)

print(merge_sort([6, 3, 8, 1, 5, 2, 7, 4]))
```

## Hash Functions for Strings

<div style="display: flex; margin-top: 0px; font-family: system-ui; line-height: 1.35; color: #212121;"><div style="width: 16px; min-width: 16px; height: 16px; margin: 9px 20px 0 0; background-color: #b0603a;"></div><div>Map each character to a number with <code>ord()</code>, add the numbers, then use <code>%</code></div></div>
<div style="display: flex; margin-top: 14px; font-family: system-ui; line-height: 1.35; color: #212121;"><div style="width: 16px; min-width: 16px; height: 16px; margin: 9px 20px 0 0; background-color: #b0603a;"></div><div><code>'cat'</code> &rarr; 99 + 97 + 116 = 312 &rarr; 312 % 10 = 2</div></div>

```python
print(ord('c'), ord('a'), ord('t'))

def hash_string(string, table_size):
    total = sum([ord(char) for char in string])
    return total % table_size

hash_string('cat', 10)
```
